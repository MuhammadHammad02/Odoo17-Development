```python
class Person():
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary

    def work(self):                    # ✅ barabar
        return f"{self.name} is working"

    def get_salary(self):              # ✅ naam badla
        return f"{self.name} salary is {self.salary}"

class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department

    def manage_department(self):       # ✅ barabar
        return f"{self.name} is managing {self.department}"

    def get_salary(self):              # ✅ naam badla
        return f"{self.name} salary is {self.salary}"

# Employee object
a = Employee("Qasim", 50000)
print(a.name)           # Qasim
print(a.salary)         # 50000
print(a.get_salary())   # Qasim salary is 50000
print(a.work())         # Qasim is working

# Manager object
b = Manager("Ali", 100000, "IT")
print(b.name)                # Ali
print(b.salary)              # 100000
print(b.department)          # IT
print(b.get_salary())        # Ali salary is 100000
print(b.manage_department()) # Ali is managing IT
```
