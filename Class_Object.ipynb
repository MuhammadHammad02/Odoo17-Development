```
class Customer:
  def __init__(self, name, mail):
    self.name = name
    self.mail = mail

customer1 = Customer('ALi', 'abc@gmail.com')
customer2 = Customer('Veli', 'xyz@gmail.com')

print(customer1.name)
print(customer1.mail)
print(customer2.name)
print(customer2.mail)
```
